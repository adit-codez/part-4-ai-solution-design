```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
from matplotlib.patches import FancyArrowPatch, FancyBboxPatch, ConnectionPatch
from matplotlib.gridspec import GridSpec
import os

BG    = '#0f1117'
CARD  = '#1a1d2e'
GRID  = '#2a2d3e'
WHITE = '#e8eaf0'
BLUE  = '#4fc3f7'
GREEN = '#66bb6a'
RED   = '#ef5350'
AMB   = '#ffa726'
PRP   = '#ab47bc'

fig, ax = plt.subplots(figsize=(20, 13))
fig.patch.set_facecolor(BG)
ax.set_facecolor(BG)
ax.set_xlim(0, 20)
ax.set_ylim(0, 13)
ax.axis('off')

# ── Helper: rounded box ───────────────────────────────────────────────────────
def rbox(ax, x, y, w, h, color, alpha=0.9, radius=0.3, lw=0, ec='none'):
    box = FancyBboxPatch((x, y), w, h,
                         boxstyle=f"round,pad={radius}",
                         facecolor=color, alpha=alpha,
                         edgecolor=ec, linewidth=lw, zorder=3)
    ax.add_patch(box)

def arrow(ax, x1, y1, x2, y2, color='#4fc3f7'):
    ax.annotate('', xy=(x2, y2), xytext=(x1, y1),
                arrowprops=dict(arrowstyle='->', color=color,
                                lw=2.2, connectionstyle='arc3,rad=0.0'),
                zorder=4)

def label(ax, x, y, txt, size=9, color=WHITE, bold=False, ha='center', va='center'):
    ax.text(x, y, txt, fontsize=size, color=color,
            fontweight='bold' if bold else 'normal',
            ha=ha, va=va, zorder=5)

# ══════════════════════════════════════════════════════════════════════════════
# TITLE BANNER
# ══════════════════════════════════════════════════════════════════════════════
rbox(ax, 0.3, 11.7, 19.4, 1.1, '#1565C0', alpha=0.95, radius=0.2)
label(ax, 10, 12.35, 'AI-Powered Customer Support Ticket Routing & Resolution', size=16, bold=True, color='white')
label(ax, 10, 11.95, 'Solution Architecture — Part 4: AI Solution Design', size=10, color='#90caf9')

# ══════════════════════════════════════════════════════════════════════════════
# LAYER 1: INPUT SOURCES
# ══════════════════════════════════════════════════════════════════════════════
label(ax, 2.4, 11.3, 'INPUT LAYER', size=8.5, color=AMB, bold=True)
sources = [('Chat\nMessages', '#0d47a1'), ('Email\nTickets', '#1a237e'),
           ('Phone\nTranscripts', '#283593'), ('Social\nMedia', '#303f9f')]
for i, (src, col) in enumerate(sources):
    bx = 0.4 + i * 1.15
    rbox(ax, bx, 10.35, 1.0, 0.75, col, radius=0.2, ec=BLUE, lw=1.2)
    label(ax, bx+0.5, 10.73, src, size=8, color=WHITE)

# ══════════════════════════════════════════════════════════════════════════════
# LAYER 2: PREPROCESSING
# ══════════════════════════════════════════════════════════════════════════════
rbox(ax, 0.3, 9.1, 4.8, 1.0, CARD, radius=0.25, ec=BLUE, lw=1.5)
label(ax, 2.7, 9.78, 'NLP PREPROCESSING PIPELINE', size=9, bold=True, color=BLUE)
steps = ['Tokenise', 'Clean', 'Remove\nstopwords', 'Encode\nsequence']
for i, s in enumerate(steps):
    bx = 0.5 + i*1.12
    rbox(ax, bx, 9.18, 1.0, 0.55, '#0d47a1', radius=0.15)
    label(ax, bx+0.5, 9.46, s, size=7.5, color=WHITE)

# arrows: input → preproc
for i in range(4):
    arrow(ax, 0.9+i*1.15, 10.35, 0.9+i*1.15, 10.12)

# ══════════════════════════════════════════════════════════════════════════════
# LAYER 3: AI MODEL BLOCK
# ══════════════════════════════════════════════════════════════════════════════
rbox(ax, 0.3, 7.0, 4.8, 1.9, CARD, radius=0.25, ec=GREEN, lw=2)
label(ax, 2.7, 8.65, 'BIDIRECTIONAL LSTM MODEL', size=9.5, bold=True, color=GREEN)

model_layers = [
    ('Embedding Layer  (vocab→64-dim vectors)', '#1b5e20'),
    ('BiLSTM Layer     (64 units each direction)', '#2e7d32'),
    ('GlobalMaxPool + Dropout(0.3)', '#388e3c'),
    ('Dense Output     (Softmax · 3 classes)', '#43a047'),
]
for i, (txt, col) in enumerate(model_layers):
    by = 8.35 - i*0.31
    rbox(ax, 0.5, by-0.13, 4.4, 0.26, col, radius=0.1)
    label(ax, 2.7, by, txt, size=7.8, color=WHITE)

# preproc → model
arrow(ax, 2.7, 9.18, 2.7, 8.9)

# ══════════════════════════════════════════════════════════════════════════════
# LAYER 4: MODEL OUTPUTS (right of model)
# ══════════════════════════════════════════════════════════════════════════════
label(ax, 7.0, 8.65, 'CLASSIFICATION\nOUTPUT', size=9, bold=True, color=GREEN)
outputs = [('POSITIVE', GREEN, 7.0, 8.1),
           ('NEUTRAL',  AMB,   7.0, 7.6),
           ('NEGATIVE', RED,   7.0, 7.1)]
for lbl, col, bx_c, by_c in outputs:
    rbox(ax, bx_c-0.85, by_c-0.18, 1.7, 0.32, col, alpha=0.3, radius=0.12, ec=col, lw=1.5)
    label(ax, bx_c, by_c, lbl, size=9, bold=True, color=col)

# model → output
arrow(ax, 5.1, 7.95, 6.15, 8.1)
arrow(ax, 5.1, 7.95, 6.15, 7.6)
arrow(ax, 5.1, 7.95, 6.15, 7.1)

# ══════════════════════════════════════════════════════════════════════════════
# LAYER 5: ROUTING ENGINE
# ══════════════════════════════════════════════════════════════════════════════
rbox(ax, 8.3, 6.9, 3.2, 1.9, CARD, radius=0.25, ec=AMB, lw=2)
label(ax, 9.9, 8.58, 'INTELLIGENT ROUTING', size=9.5, bold=True, color=AMB)
routing_rules = [
    'Urgency Score  ──  flag high-priority',
    'Sentiment      ──  route to specialist',
    'Topic cluster  ──  auto-assign team',
    'SLA timer      ──  escalate if needed',
]
for i, rule in enumerate(routing_rules):
    label(ax, 9.9, 8.25 - i*0.29, rule, size=7.8, color=WHITE)

# output → routing
arrow(ax, 7.9, 7.6, 8.3, 7.7)

# ══════════════════════════════════════════════════════════════════════════════
# LAYER 6: ACTION OUTPUTS
# ══════════════════════════════════════════════════════════════════════════════
actions = [
    ('Auto-Respond\n(Templates)',  BLUE,  12.1, 8.4),
    ('Agent Queue\n(Escalated)',   AMB,   12.1, 7.8),
    ('Knowledge Base\n(Self-serve)', GREEN, 12.1, 7.2),
    ('Supervisor\n(Critical)',     RED,   12.1, 6.6),
]
label(ax, 13.0, 8.75, 'ACTION LAYER', size=8.5, color=AMB, bold=True)
for lbl, col, bx, by in actions:
    rbox(ax, bx-0.7, by-0.23, 1.4, 0.42, col, alpha=0.25, radius=0.15, ec=col, lw=1.5)
    label(ax, bx, by, lbl, size=7.5, color=col, bold=True)
    arrow(ax, 11.5, 7.7, bx-0.7, by)

# ══════════════════════════════════════════════════════════════════════════════
# LAYER 7: FEEDBACK LOOP
# ══════════════════════════════════════════════════════════════════════════════
rbox(ax, 0.3, 5.8, 4.8, 0.95, CARD, radius=0.2, ec=PRP, lw=1.5)
label(ax, 2.7, 6.5, 'FEEDBACK & CONTINUOUS LEARNING LOOP', size=8.5, bold=True, color=PRP)
fb_items = ['Agent outcome labels', 'CSAT scores', 'Resolution time', 'Model re-training']
for i, f in enumerate(fb_items):
    rbox(ax, 0.5+i*1.12, 5.88, 1.05, 0.4, '#4a148c', radius=0.1, alpha=0.8)
    label(ax, 0.5+i*1.12+0.525, 6.08, f, size=7, color=WHITE)

# model → feedback (curved)
ax.annotate('', xy=(2.7, 6.75), xytext=(2.7, 6.95),
            arrowprops=dict(arrowstyle='->', color=PRP, lw=1.8), zorder=4)

# feedback → model (loop back)
ax.annotate('', xy=(0.3, 7.4), xytext=(0.3, 6.28),
            arrowprops=dict(arrowstyle='->', color=PRP, lw=1.5,
                            connectionstyle='arc3,rad=-0.4'), zorder=4)
label(ax, 0.05, 6.9, 'Re-train', size=7, color=PRP, ha='center')

# ══════════════════════════════════════════════════════════════════════════════
# METRICS PANEL (bottom right)
# ══════════════════════════════════════════════════════════════════════════════
rbox(ax, 8.3, 4.5, 7.0, 3.2, CARD, radius=0.25, ec='#546e7a', lw=1.5)
label(ax, 11.8, 7.45, 'BUSINESS IMPACT — PROJECTED METRICS', size=10, bold=True, color=WHITE)

metrics = [
    ('Manual Hours',       '509 hrs/mo',  '200 hrs/mo',  '-61%',  RED,   GREEN),
    ('Resolution Time',    '35.2 hrs',    '8.5 hrs',     '-76%',  RED,   GREEN),
    ('Error Rate',         '8.9%',        '<3%',         '-66%',  RED,   GREEN),
    ('CSAT Score',         '6.7 / 10',    '8.5 / 10',    '+27%',  AMB,   GREEN),
    ('Cases Handled / mo', '2,800',       '5,000+',      '+78%',  BLUE,  GREEN),
]
col_x = [8.6, 10.5, 12.5, 14.3, 15.7]
headers = ['Metric', 'Current', 'Projected', 'Change', '']
for i, (h, cx) in enumerate(zip(headers, col_x)):
    label(ax, cx, 7.05, h, size=8, bold=True, color='#90a4ae', ha='left')

for row_i, (metric, curr, proj, chg, c_col, p_col) in enumerate(metrics):
    ry = 6.65 - row_i * 0.42
    if row_i % 2 == 0:
        rbox(ax, 8.4, ry-0.18, 6.8, 0.38, '#212535', radius=0.08, alpha=0.6)
    label(ax, 8.6, ry, metric, size=8.2, color=WHITE, ha='left', bold=True)
    label(ax, 10.5, ry, curr, size=8.2, color=c_col, ha='left')
    label(ax, 12.5, ry, proj, size=8.2, color=p_col, ha='left')
    rbox(ax, 14.25, ry-0.13, 1.1, 0.28, p_col, alpha=0.25, radius=0.08)
    label(ax, 14.8, ry, chg, size=8.5, color=p_col, ha='center', bold=True)

# ══════════════════════════════════════════════════════════════════════════════
# DATA SOURCES BOX (bottom left)
# ══════════════════════════════════════════════════════════════════════════════
rbox(ax, 0.3, 4.5, 7.6, 1.1, CARD, radius=0.2, ec='#546e7a', lw=1.2)
label(ax, 4.1, 5.35, 'DATA PIPELINE', size=9, bold=True, color='#90a4ae')
data_items = ['Historical tickets\n(1,500+ records)', 'Agent resolution\nlabels',
              'CSAT survey\nresponses', 'Channel metadata\n& timestamps',
              'NLP features\n(TF-IDF / embeddings)', 'Operational KPIs\n(time, error rate)']
for i, d in enumerate(data_items):
    bx = 0.5 + i*1.22
    rbox(ax, bx, 4.58, 1.1, 0.65, '#1e3a5f', radius=0.12, ec='#4fc3f7', lw=0.8)
    label(ax, bx+0.55, 4.9, d, size=6.8, color=WHITE)

# ══════════════════════════════════════════════════════════════════════════════
# RESPONSIBLE AI BOX (far right strip)
# ══════════════════════════════════════════════════════════════════════════════
rbox(ax, 15.8, 4.5, 3.9, 8.3, '#1a1a2e', radius=0.2, ec='#ef5350', lw=1.5)
label(ax, 17.75, 12.55, 'RESPONSIBLE AI', size=9, bold=True, color=RED)
label(ax, 17.75, 12.2, 'SAFEGUARDS', size=9, bold=True, color=RED)
rai = [
    ('Human-in-loop review', RED),
    ('Confidence threshold', AMB),
    ('Bias audit quarterly', AMB),
    ('Data privacy (GDPR)', BLUE),
    ('Model explainability', GREEN),
    ('Fallback to agent', GREEN),
    ('Audit trail logging', BLUE),
    ('No auto-close tickets', AMB),
    ('Regular retraining', GREEN),
    ('Stakeholder alerts', RED),
]
for i, (item, col) in enumerate(rai):
    label(ax, 16.1, 11.75-i*0.52, f'  • {item}', size=8, color=col, ha='left')

# Separator line
ax.plot([15.8, 15.8], [4.4, 12.85], color='#ef5350', linewidth=1.5, alpha=0.4, zorder=2)

# ══════════════════════════════════════════════════════════════════════════════
# LEGEND
# ══════════════════════════════════════════════════════════════════════════════
legend_items = [('Input / Data', BLUE), ('AI Model', GREEN),
                ('Business Logic', AMB), ('Risk Controls', RED), ('Feedback', PRP)]
for i, (lbl2, col) in enumerate(legend_items):
    bx2 = 8.5 + i * 1.42
    rbox(ax, bx2, 4.05, 1.3, 0.3, col, alpha=0.3, radius=0.08, ec=col, lw=1.2)
    label(ax, bx2+0.65, 4.2, lbl2, size=7.5, color=col)

# Subtitle / footer
label(ax, 10, 0.3, 'Part 4 · AI Solution Design · Customer Support Domain · Bidirectional LSTM + Intelligent Routing Engine',
      size=8, color='#546e7a')

output_path = '/home/claude/part-4-ai-solution-design/diagrams/solution_architecture.png'
os.makedirs(os.path.dirname(output_path), exist_ok=True)
plt.savefig(output_path,
            dpi=150, bbox_inches='tight', facecolor=BG)
plt.close()
print("Architecture diagram saved")
```
